# CaptionAI: Model Evaluation & Metrics
Evaluation of generated captions against human ground-truth reference captions using standard natural language metrics:
* **BLEU-1** (Unigram Precision)
* **BLEU-2** (Bigram Precision)
* **BLEU-3** (Trigram Precision)
* **BLEU-4** (4-gram Precision)
* **METEOR** (Harmonic mean of precision and recall with unigram alignment)

```python
from ml.evaluation.metrics import compute_bleu_scores, compute_meteor, evaluate_caption_batch

references = [
    ['a person standing on the beach during a beautiful golden sunset', 'a man watching the sunset by the ocean'],
    ['a happy dog running in a sunny green park', 'a brown dog playing on the grass']
]
candidates = [
    'a person standing on the beach during a beautiful golden sunset',
    'a dog running happily in the green grass'
]

results = evaluate_caption_batch(references, candidates)
print('=== Benchmark Evaluation Results ===')
for k, v in results.items():
    print(f'{k}: {v}')
```
