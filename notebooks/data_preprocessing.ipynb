# CaptionAI: Data Preprocessing Pipeline
This notebook details the image and text preprocessing steps for training the ResNet50 + LSTM Image Captioning model on Flickr8k.

```python
import os
import re
from PIL import Image
import matplotlib.pyplot as plt
from ml.preprocessing.text_tokenizer import Vocabulary
from ml.preprocessing.image_transforms import get_image_transforms

print('Libraries successfully imported!')
```

## 1. Caption Cleaning & Vocabulary Construction
Text captions must be cleaned (lowercased, stripped of non-alphanumeric punctuation) and tokenized with special tokens: `<pad>`, `<start>`, `<end>`, and `<unk>`.

```python
sample_captions = [
    'A person standing on the beach during a golden sunset.',
    'A dog running happily across green grass in a park!',
    'A woman drinking hot coffee at a modern cafe.',
    'A group of friends laughing together outdoors.'
]

vocab = Vocabulary(freq_threshold=1)
vocab.build_vocab(sample_captions)
print(f'Vocabulary size: {len(vocab)}')
print('Sample tokens in vocab:', list(vocab.stoi.items())[:15])

test_sentence = 'A person enjoying the golden sunset.'
numericalized = vocab.numericalize(test_sentence)
print(f'Numericalized indices: {numericalized}')
decoded = vocab.decode(numericalized)
print(f'Decoded back: {decoded}')
```

## 2. Image Preprocessing & ResNet Normalization
Images are resized to 256x256, cropped to 224x224, converted to PyTorch tensors, and normalized with ImageNet mean `[0.485, 0.456, 0.406]` and standard deviation `[0.229, 0.224, 0.225]`.

```python
transform = get_image_transforms(is_train=False)
dummy_img = Image.new('RGB', (300, 300), color=(100, 150, 200))
tensor_img = transform(dummy_img)
print(f'Processed Tensor shape: {tensor_img.shape}')
print(f'Tensor min: {tensor_img.min():.2f}, max: {tensor_img.max():.2f}')
```
