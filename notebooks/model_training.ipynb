# CaptionAI: Model Training Pipeline
Architecture: **ResNet50 CNN Encoder + LSTM Decoder**

Pipeline Flow:
1. ResNet50 extracts a 2048-dim feature vector from the 224x224 image.
2. A linear projection aligns the visual feature with the decoder embedding dimension (512-dim).
3. The LSTM decoder receives the visual feature followed by word token embeddings.
4. Cross-entropy loss trains the model with teacher forcing.

```python
import torch
import torch.nn as nn
from ml.models.caption_model import ImageCaptioningModel
from ml.preprocessing.text_tokenizer import Vocabulary

device = 'cuda' if torch.cuda.is_available() else ('mps' if torch.backends.mps.is_available() else 'cpu')
print(f'Training on device: {device}')
```

```python
vocab = Vocabulary.load('checkpoints/vocab.json')
model = ImageCaptioningModel(
    embed_dim=512,
    hidden_dim=512,
    vocab_size=len(vocab),
    fine_tune_encoder=False
).to(device)

print('Model Architecture:')
print(model)
```

## Training Step (Loss & Optimizer)
We use Adam optimizer (`lr=3e-4`) and `CrossEntropyLoss(ignore_index=pad_token)`.

```python
criterion = nn.CrossEntropyLoss(ignore_index=vocab.stoi[vocab.PAD_TOKEN])
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4)
print('Criterion and optimizer configured successfully.')
```
